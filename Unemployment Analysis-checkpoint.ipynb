```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
%matplotlib inline
```

```python
# Cell 2: Load data
# Make sure the CSV is in the same folder as this notebook
# Common name: Unemployment_in_India.csv (download from https://www.kaggle.com/datasets/gokulrajkmv/unemployment-in-india)

df = pd.read_csv('unemployment in india.csv')
print("Shape:", df.shape)
print("\nColumns:", df.columns.tolist())
print("\nFirst 5 rows:\n", df.head())
print("\nInfo:")
df.info()
print("\nMissing values:\n", df.isnull().sum())
```

```python
# Cell 3: Clean the data
# Drop missing rows if any (common in these datasets)
df = df.dropna()

# Fix column names (they often have spaces/extra chars)
df.columns = df.columns.str.strip()  # remove leading/trailing spaces

# Convert 'Date' to datetime (format may vary: dd-mm-yyyy)
df['Date'] = pd.to_datetime(df['Date'], dayfirst=True)  # or format='%d-%m-%Y'

# Sort by date
df = df.sort_values('Date')

print("Cleaned data shape:", df.shape)
print(df['Date'].min(), "to", df['Date'].max())  # Check COVID period (March 2020+)
```

```python
# Cell 4: Basic Exploration
print("Average Unemployment Rate:", df['Estimated Unemployment Rate (%)'].mean())
print("\nUnemployment by State (mean):\n", df.groupby('Region')['Estimated Unemployment Rate (%)'].mean().sort_values(ascending=False))

# Overall trend
plt.figure(figsize=(12,6))
sns.lineplot(x='Date', y='Estimated Unemployment Rate (%)', data=df, errorbar=None)
plt.title('Unemployment Rate Trend Over Time (2019-2020)')
plt.axvline(pd.to_datetime('2020-03-25'), color='red', linestyle='--', label='COVID Lockdown Start (India)')
plt.legend()
plt.show()
```

```python
# Cell 5: COVID Impact Analysis
# Filter pre-COVID (before March 2020) and during/post
pre_covid = df[df['Date'] < '2020-03-01']
during_covid = df[df['Date'] >= '2020-03-01']

print("Pre-COVID Avg Unemployment:", pre_covid['Estimated Unemployment Rate (%)'].mean())
print("During/Post-COVID Avg Unemployment:", during_covid['Estimated Unemployment Rate (%)'].mean())
print(f"Increase: {during_covid['Estimated Unemployment Rate (%)'].mean() - pre_covid['Estimated Unemployment Rate (%)'].mean():.2f}%")

# State-wise impact
impact = during_covid.groupby('Region')['Estimated Unemployment Rate (%)'].mean() - pre_covid.groupby('Region')['Estimated Unemployment Rate (%)'].mean()
impact = impact.sort_values(ascending=False)

plt.figure(figsize=(12,8))
impact.plot(kind='bar')
plt.title('Increase in Unemployment Rate by State (COVID Impact)')
plt.ylabel('Percentage Point Increase')
plt.xticks(rotation=90)
plt.show()
```

```python
# Cell 6: Seasonal / Monthly Trends
df['Month'] = df['Date'].dt.month
df['Year'] = df['Date'].dt.year

monthly_avg = df.groupby(['Year', 'Month'])['Estimated Unemployment Rate (%)'].mean().reset_index()

plt.figure(figsize=(10,6))
sns.lineplot(x='Month', y='Estimated Unemployment Rate (%)', hue='Year', data=monthly_avg, marker='o')
plt.title('Monthly Unemployment Trends (Seasonality)')
plt.xticks(range(1,13))
plt.show()
```

```python
# Aggregate the data using the correct column name
sunburst_df = df.groupby(['Area', 'Region'])['Estimated Unemployment Rate (%)'].mean().reset_index()

# Create the chart
fig = px.sunburst(
    sunburst_df, 
    path=['Area', 'Region'], 
    values='Estimated Unemployment Rate (%)',
    color='Estimated Unemployment Rate (%)',
    color_continuous_scale='RdYlGn_r',
    title='Average Unemployment Rate by Area and Region'
)
fig.show()
```
